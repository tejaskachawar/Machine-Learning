```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_moons
# With adding noise   noise is used to make data points more scatter  
x,y = make_moons(n_samples=250,noise=0.05)
# x contains two features, so I create two DataFrame columns
df = {"data1":x[:,0],"data2":x[:,1]}
dataset  = pd.DataFrame(df) # passing here whole dataset as it don't have output column
# dataset['predict'] = db.fit_predict(dataset[['data1', 'data2']]) # for specific features
dataset.head(3)
```

```python
# To see the distribution of data points
sns.scatterplot(x=dataset.data1,y=dataset.data2)
plt.show()
```

```python
from sklearn.cluster import DBSCAN
db = DBSCAN(eps=0.2,min_samples=5)
dataset['predict'] = db.fit_predict(dataset)
# Here i have stored prediction in predict column
```

```python
# To see prediction plotting scatter plot here with predict as hue parameter
sns.scatterplot(x="data1", y="data2",data=dataset, hue="predict")
plt.show()
```

```python
# To see original prediction
sns.scatterplot(x="data1", y="data2",data=dataset, hue=y)
plt.show()
```
