Feature and target segregation

```python
# dropping Loan ID
# Remove LoanID because it is an identifier rather than a useful predictive feature.
df.drop('LoanID', axis = 1 , inplace = True)
```

```python
# Create X containing all predictor features by removing the target column Default.
x = df.drop('Default', axis = 1)
# Create y containing the target variable Default that the models will predict.
y = df['Default']
```

```python
from sklearn.preprocessing import LabelEncoder
# filtering only non numeric col
# Select only object/string columns so categorical features can be encoded.
df_str = df.select_dtypes(include='str',exclude='float64')
df_str
```

```python
# Create a LabelEncoder object for converting categorical labels into integer codes.
enc = LabelEncoder()
# Loop through every categorical column.
for cols in df_str:
    # Fit the encoder on the categorical column and replace its categories with integer codes.
    # Here we using x because we earlier stored all cols except default in X

    x[cols] = enc.fit_transform(x[cols])

x
```

Standardization

```python
# Import the required objects from sklearn.preprocessing.
from sklearn.preprocessing import MinMaxScaler
# Create a MinMaxScaler that maps each feature to the range 0 to 1.
scaler = MinMaxScaler(feature_range=(0, 1))
```

```python
# Fit the scaler on all feature columns and transform them to the 0–1 range.
x[x.columns] = scaler.fit_transform(x[x.columns])

# Display the first rows after scaling to verify the transformed values.
x.head(8)
```
